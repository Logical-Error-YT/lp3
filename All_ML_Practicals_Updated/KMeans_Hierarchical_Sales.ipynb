## 1. Load and select numeric features

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans, AgglomerativeClustering
from sklearn.metrics import silhouette_score

df = pd.read_csv("sales_data_sample.csv", encoding="latin1")
features = ["QUANTITYORDERED", "PRICEEACH", "SALES", "MSRP"]
data = df[features].dropna()
data.head()
```

## 2. Standardize features

```python
scaler = StandardScaler()
X = scaler.fit_transform(data)
print("Rows used:", X.shape[0])
```

## 3. Elbow method for K-Means

```python
inertia = []
for k in range(1, 11):
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    km.fit(X)
    inertia.append(km.inertia_)

plt.plot(range(1, 11), inertia, marker="o")
plt.xlabel("Number of clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Method")
plt.grid()
plt.show()
```

## 4. K-Means clustering (choose K = 3 for demonstration)

```python
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
data["KMeans_Cluster"] = kmeans.fit_predict(X)
print(data["KMeans_Cluster"].value_counts().sort_index())
print("Silhouette Score:", silhouette_score(X, data["KMeans_Cluster"]))
data.head()
```

## 5. Hierarchical clustering

```python
hier = AgglomerativeClustering(n_clusters=3, linkage="ward")
data["Hierarchical_Cluster"] = hier.fit_predict(X)
print(data["Hierarchical_Cluster"].value_counts().sort_index())
data.head()
```

## 6. Visualize clusters (first two standardized features)

```python
plt.scatter(X[:,0], X[:,1], c=data["KMeans_Cluster"], s=15)
plt.xlabel(features[0] + " (scaled)")
plt.ylabel(features[1] + " (scaled)")
plt.title("K-Means Clusters")
plt.show()
```
