## Gradient Descent: Minimize y = (x + 3)^2 starting from x = 2

```python
import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return (x + 3)**2

def derivative(x):
    return 2*(x + 3)

x = 2.0
learning_rate = 0.1
iterations = 30
history = []

for i in range(iterations):
    history.append([i, x, f(x)])
    x = x - learning_rate * derivative(x)

print("Final x:", x)
print("Minimum y:", f(x))
print("Expected minimum: x = -3, y = 0")
```

## Plot convergence

```python
history = np.array(history)
plt.plot(history[:,0], history[:,1], marker="o")
plt.xlabel("Iteration")
plt.ylabel("x")
plt.title("Gradient Descent Convergence")
plt.grid()
plt.show()
```
