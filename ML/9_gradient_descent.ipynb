# Gradient Descent for y = (x + 3)^2

Starting at x = 2, update x using the derivative and a learning rate until convergence. The minimum is at x = -3.

```python
import pandas as pd
x, lr = 2.0, 0.1
history = []
for i in range(100):
    y = (x + 3)**2
    history.append((i, x, y))
    x -= lr * 2*(x + 3)
print(pd.DataFrame(history, columns=["iteration", "x", "y"]).head(10))
print(f"Local minimum: x={x:.6f}, y={(x+3)**2:.10f}")
```

```python
import matplotlib.pyplot as plt
xs = [r[1] for r in history]
plt.plot(range(len(xs)), xs, marker=".")
plt.xlabel("Iteration"); plt.ylabel("x"); plt.title("Gradient Descent Convergence"); plt.grid(True); plt.show()
```
