# Email Spam Detection: Binary Classification

Place `spam.csv` in this folder. The notebook accepts common dataset columns such as `text`/`message` and `label`/`v1`; labels should identify spam and not-spam.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

df = pd.read_csv("spam.csv", encoding="latin-1")
text_col = next(c for c in ["text", "message", "v2", "email"] if c in df.columns)
label_col = next(c for c in ["label", "v1", "category", "target"] if c in df.columns)
df = df[[text_col, label_col]].dropna()
X = df[text_col].astype(str)
y = df[label_col].astype(str).str.lower().map(lambda x: 1 if x in ["spam", "1", "yes"] else 0)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=.2, random_state=42, stratify=y)
model = make_pipeline(TfidfVectorizer(stop_words="english"), MultinomialNB())
model.fit(X_train, y_train)
pred = model.predict(X_test)
print("Accuracy:", accuracy_score(y_test, pred))
print(classification_report(y_test, pred, target_names=["NotSpam", "Spam"]))
print("Confusion matrix:\n", confusion_matrix(y_test, pred))
sample = ["Congratulations! You won a free prize. Click now."]
print("Sample:", "Spam" if model.predict(sample)[0] else "NotSpam")
```
