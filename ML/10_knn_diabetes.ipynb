# K-Nearest Neighbors on diabetes.csv

Place `diabetes.csv` in this folder. The target column is usually `Outcome` (0 = non-diabetic, 1 = diabetic).

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

df = pd.read_csv("diabetes.csv")
target = "Outcome" if "Outcome" in df.columns else df.columns[-1]
X, y = df.drop(columns=target), df[target]
# Common diabetes datasets use zero to represent missing values in these measurements
for c in ["Glucose", "BloodPressure", "SkinThickness", "Insulin", "BMI"]:
    if c in X.columns: X[c] = X[c].replace(0, float("nan"))
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=.2, random_state=42, stratify=y)
model = make_pipeline(SimpleImputer(strategy="median"), StandardScaler(), KNeighborsClassifier(n_neighbors=5))
model.fit(X_train, y_train)
pred = model.predict(X_test)
print("Accuracy:", accuracy_score(y_test, pred))
print(classification_report(y_test, pred))
print("Confusion matrix:\n", confusion_matrix(y_test, pred))
```
